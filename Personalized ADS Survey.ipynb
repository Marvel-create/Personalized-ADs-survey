**Client Brief for Data Analysis Project**

**Project Title:
Analysis of Data-Driven Advertising Amongst University Students**

**Objective**:
***The goal of this analysis is to understand university students' interaction with personalized ads, their awareness of ad tracking, and their general sentiment towards data-driven advertising. The insights from this study will help advertisers, businesses, and policymakers tailor their approaches to digital marketing for a university audience.***

**Key Research Questions:**
*Ad Awareness & Engagement*:

1. How often do students notice targeted advertisements?
2. Which platforms deliver the most personalized ads to them?
3. Do students find personalized ads relevant and engaging?

**Behavioral Insights:**

1. How much time do students spend online daily?
2. Are students more likely to purchase from personalized ads?
3. What is their general experience with online ads (positive, neutral, negative)?

**Privacy & Data Tracking Concerns:**

1. How comfortable are students with ad tracking?
2. Do they actively change privacy settings to limit tracking?
3. What percentage prefer non-tracking ads?

**Actionable Recommendations:**

1. Based on the data, what strategies can businesses use to improve ad relevance?
2. Should advertisers change their approach to personalization to align with student preferences?
3. Are students becoming more privacy-conscious, and how should advertisers respond?

**Expected Deliverables:**

1. Data Cleaning & Preprocessing (Ensure data accuracy and structure)
2. Descriptive Statistics (Summarize key metrics)
3. Visualizations & Insights (Graphs, charts, and trends)
4. Correlation & Segmentation Analysis (Finding relationships between different responses)
5. Privacy & Ad Engagement Trends (Identifying key takeaways for marketers)
6. Final Report with Business Recommendations

Now to Answer the second qustion
Which platforms deliver the most personalized ads to them?


```python
platform_counts = survey["Personalized Ad Platforms"].value_counts()
print(platform_counts)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
sns.barplot(x=platform_counts.index, y=platform_counts.values, palette="viridis")
plt.xlabel("Platform")
plt.ylabel("Number of Responses")
plt.title("Platforms Delivering the Most Personalized Ads")
plt.xticks(rotation=45)  # Rotate labels if needed
plt.show()
```

**THE ABOVE VISUALIZATION IS TOO CONFUSING BECASE THE ANSWERS ARE STILL TOGETHER**

```python
#To solve the above problem, we can split the answers and then convert it to a new dataframe
from collections import Counter

# Flatten the list by splitting each response on ", " and counting occurrences
all_platforms = survey["Personalized Ad Platforms"].dropna().str.split(", ").sum()

# Count occurrences of each platform
platform_counts = Counter(all_platforms)

# Convert to a DataFrame
platform_df = pd.DataFrame(platform_counts.items(), columns=["Platform", "Count"]).sort_values(by="Count", ascending=False)

print(platform_df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
sns.barplot(x=platform_df["Platform"], y=platform_df["Count"], palette="magma")
plt.xlabel("Platform")
plt.ylabel("Number of Mentions")
plt.title("Most Mentioned Personalized Ad Platforms")
plt.xticks(rotation=45)
plt.show()
```
